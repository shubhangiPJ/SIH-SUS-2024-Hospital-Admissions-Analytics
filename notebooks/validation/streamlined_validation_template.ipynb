# SIH/SUS Streamlined Monthly Validation — Final Template

Used for April–December 2024, following the decision gate reached after Jan–Mar full validation.

**Structure:**
- Layer 1 — Structural conformance (always run)
- Layer 2 — Known recurring quality metrics (always run, compared to Jan–Mar baseline range)
- Layer 3 — Change detection (always run, but only *prints* when something deviates from the Jan–Mar baseline)

All baseline values below (dtypes, missingness, stable fields) were extracted directly from the actual Jan/Feb/Mar notebook outputs — not assumed.

**Escalation rule:** if Layer 1 fails, or Layer 3 flags anything, stop and pull the relevant domain section from the full validation notebook (`01_validation_template.ipynb`) before proceeding.

## Layer 2 — Known Recurring Quality Metrics
Compare each result to the Jan–Mar range noted in `diff_log.md`.

### 2.1 Date conversions (needed below)

```
df['NASC_dt'] = pd.to_datetime(df['NASC'].astype(str), format='%Y%m%d')
df['DT_INTER_dt'] = pd.to_datetime(df['DT_INTER'].astype(str), format='%Y%m%d')
df['DT_SAIDA_dt'] = pd.to_datetime(df['DT_SAIDA'].astype(str), format='%Y%m%d')
```

### 2.2 Implausible age (>120 yrs)
Baseline: Jan=11, Feb=19, Mar=16

```
implausible_age = ((df['DT_INTER_dt'] - df['NASC_dt']).dt.days / 365.25 > 120).sum()
AGE_BASELINE_MIN, AGE_BASELINE_MAX = 11, 19

print(f"Implausible age records: {implausible_age}  (baseline range: {AGE_BASELINE_MIN}-{AGE_BASELINE_MAX})")
if AGE_BASELINE_MIN <= implausible_age <= AGE_BASELINE_MAX:
    print("Status: WITHIN baseline")
else:
    print("\u26a0 Status: OUTSIDE baseline")
```

### 2.3 DIAS_PERM vs. calendar-day mismatch
Baseline rate: 0.94%–1.01%

```
mismatch = ((df['DT_SAIDA_dt'] - df['DT_INTER_dt']).dt.days != df['DIAS_PERM']).sum()
mismatch_rate = mismatch / len(df) * 100
DIAS_BASELINE_MIN, DIAS_BASELINE_MAX, BUFFER = 0.94, 1.01, 0.5

print(f"DIAS_PERM mismatches: {mismatch:,} ({mismatch_rate:.2f}%)  (baseline range: {DIAS_BASELINE_MIN}%-{DIAS_BASELINE_MAX}%)")
if (DIAS_BASELINE_MIN - BUFFER) <= mismatch_rate <= (DIAS_BASELINE_MAX + BUFFER):
    print("Status: WITHIN baseline")
else:
    print("\u26a0 Status: OUTSIDE baseline")
```

### 2.4 CEP formatting (non-8-digit)
Baseline rate: 8.26%–8.52%

```
cep_bad = (df['CEP'].astype(str).str.len() != 8).sum()
cep_rate = cep_bad / len(df) * 100
CEP_BASELINE_MIN, CEP_BASELINE_MAX, BUFFER = 8.26, 8.52, 0.5

print(f"Non-8-digit CEP records: {cep_bad:,} ({cep_rate:.2f}%)  (baseline range: {CEP_BASELINE_MIN}%-{CEP_BASELINE_MAX}%)")
if (CEP_BASELINE_MIN - BUFFER) <= cep_rate <= (CEP_BASELINE_MAX + BUFFER):
    print("Status: WITHIN baseline")
else:
    print("\u26a0 Status: OUTSIDE baseline")
```

### 2.5 RACA_COR = 99 and SEXO codes
Baseline RACA_COR=99: 258–282. SEXO should show only {1, 3}.

```
raca_99 = (df['RACA_COR'] == 99).sum()
sexo_codes = sorted(df['SEXO'].unique())
RACA_BASELINE_MIN, RACA_BASELINE_MAX = 258, 282
SEXO_BASELINE = [1, 3]

print(f"RACA_COR = 99 count: {raca_99}  (baseline range: {RACA_BASELINE_MIN}-{RACA_BASELINE_MAX})")
print("Status:", "WITHIN baseline" if RACA_BASELINE_MIN <= raca_99 <= RACA_BASELINE_MAX else "\u26a0 OUTSIDE baseline")

print(f"SEXO observed codes: {sexo_codes}  (baseline: {SEXO_BASELINE})")
print("Status:", "WITHIN baseline" if sexo_codes == SEXO_BASELINE else "\u26a0 OUTSIDE baseline - new/missing code")
```

## Layer 3 — Change Detection
These checks only print output when something **deviates** from the verified Jan–Mar baseline. Silence = no change = nothing to review.

### 3.1 Stable always-empty fields
Confirmed 100% missing in all of Jan, Feb, Mar: `NUM_PROC`, `INFEHOSP`, `GESTOR_DT`, `DIAGSEC8`, `DIAGSEC9`. Flags if any of these suddenly contain data — that would mean SIH/SUS started populating a previously unused field.

```
STABLE_EMPTY_FIELDS = ['NUM_PROC', 'INFEHOSP', 'GESTOR_DT', 'DIAGSEC8', 'DIAGSEC9']

empty_field_changes = {}
for col in STABLE_EMPTY_FIELDS:
    pct_missing = df[col].isna().mean() * 100
    if pct_missing < 100:
        empty_field_changes[col] = pct_missing

if empty_field_changes:
    print("⚠ PREVIOUSLY EMPTY FIELDS NOW CONTAIN DATA:")
    for col, pct in empty_field_changes.items():
        print(f"  {col}: now {100-pct:.2f}% populated")
else:
    print("All 5 stable-empty fields remain 100% missing. No change.")
```

### 3.2 Stable constant fields
Confirmed cardinality=1 (constant value) in all of Jan, Feb, Mar — mostly unused financial sub-components plus `NATUREZA`. Flags if any of these start varying.

```
STABLE_CONSTANT_FIELDS = ['NATUREZA', 'VAL_SADT', 'VAL_RN', 'VAL_ACOMP', 'VAL_ORTP', 'VAL_SANGUE', 'VAL_SADTSR', 'VAL_TRANSP', 'VAL_OBSANG', 'VAL_PED1AC', 'RUBRICA', 'TOT_PT_SP']

constant_field_changes = {}
for col in STABLE_CONSTANT_FIELDS:
    n_unique = df[col].nunique()
    if n_unique > 1:
        constant_field_changes[col] = n_unique

if constant_field_changes:
    print("⚠ PREVIOUSLY CONSTANT FIELDS NOW VARY:")
    for col, n in constant_field_changes.items():
        print(f"  {col}: now has {n} unique values")
else:
    print("All 12 stable-constant fields remain constant. No change.")
```
